```python
import os
os.chdir('/Users/malikfarouh/Documents/ML workspace')

import xgboost
import numpy as np
import ROOT

# Load your saved booster
clf = xgboost.XGBClassifier()
clf.load_model("mu2e_project/trkQual_model.json")

# Save for TMVA RBDT
ROOT.TMVA.Experimental.SaveXGBoost(clf, "myBDT", "mu2e_project/trkQual_bdt.root", num_inputs=7)
print("Saved successfully!")

# Verify scores match
bdt = ROOT.TMVA.Experimental.RBDT("myBDT", "mu2e_project/trkQual_bdt.root")
test_input = np.zeros(7, dtype=np.float32)
score_rbdt = bdt.Compute(test_input)
score_xgb = clf.predict_proba(np.zeros((1,7), dtype=np.float32))[:,1]
print(f"RBDT score:    {score_rbdt}")
print(f"XGBoost score: {score_xgb[0]:.6f}")
```
